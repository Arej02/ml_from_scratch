```python
import numpy as np
```

```python
inputs=[
    [1,2,3,2.5],
    [2,5,-1,2],
    [-1.5,2.7,3.3,-0.8]
]
weights=[
    [0.2,0.8,-0.5,1],
    [0.5,-0.91,0.26,-0.5],
    [-0.26,-0.27,0.17,0.87]
]
bias=[2,3,0.5]

output=np.dot(inputs,np.array(weights).T)+bias
print(output)
```
